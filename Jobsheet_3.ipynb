##Praktikum 1

```python
class BankAccount:
    def __init__(self, owner, balance):
        # Atribut private
        self.__owner = owner
        self.__balance = balance

    def deposit(self, amount):
        """Method untuk menambahkan saldo."""
        if amount > 0:
            self.__balance += amount
            print(f"{amount} telah ditambahkan ke akun {self.__owner}.")
        else:
            print("Jumlah deposit harus lebih dari 0.")

    def withdraw(self, amount):
        """Method untuk menarik saldo."""
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"{amount} telah ditarik dari akun {self.__owner}.")
        else:
            print("Saldo tidak mencukupi.")

    def get_balance(self):
        """Method untuk mendapatkan saldo."""
        return self.__balance


# Contoh penggunaan
if __name__ == "__main__":
    # Membuat objek
    alice_account = BankAccount(owner="Alice", balance=1000)

    # Deposit uang
    alice_account.deposit(500)     # Berhasil
    alice_account.deposit(-100)    # Gagal

    # Withdraw uang
    alice_account.withdraw(300)    # Berhasil
    alice_account.withdraw(2000)   # Gagal

    # Ambil saldo
    current_balance = alice_account.get_balance()
    print(f"Saldo terakhir di akun Alice: {current_balance}")

    # Akses private (tidak disarankan)
    # print(alice_account.__balance)  # ERROR
    # Cara Python simpan:
    # print(alice_account._BankAccount__balance)
```

##Praktikum 2

```python
class Employee:
    def __init__(self, name, salary):
        # Atribut privat
        self.__name = name
        self.__salary = salary

    def update_salary(self, increase):
        """Method untuk menaikkan gaji dengan validasi."""
        if increase > 0:
            self.__salary += increase
            print(f"Gaji telah dinaikkan sebesar {increase}.")
        else:
            print("Nilai kenaikan harus lebih dari 0.")

    def set_salary(self, new_salary):
        """Method untuk mengubah gaji dengan validasi."""
        if new_salary >= 0:
            self.__salary = new_salary
            print(f"Gaji diatur ulang menjadi {new_salary}.")
        else:
            print("Gaji tidak dapat bernilai negatif.")

    def get_salary(self):
        """Method untuk mendapatkan informasi gaji."""
        return self.__salary

    def get_employee_info(self):
        """Method untuk menampilkan informasi karyawan."""
        return f"Employee: {self.__name}, Gaji: {self.__salary}"


# Contoh penggunaan
if __name__ == "__main__":
    # Membuat objek
    employee1 = Employee("John Doe", 50000)

    # Tampilkan info
    print(employee1.get_employee_info())

    # Update gaji
    employee1.update_salary(5000)
    print(f"Gaji setelah kenaikan: {employee1.get_salary()}")

    # Set gaji baru
    employee1.set_salary(60000)
    print(f"Informasi terbaru: {employee1.get_employee_info()}")

    # Akses langsung (tidak disarankan)
    # print(employee1.__salary)  # ERROR
    # print(employee1._Employee__salary)  # Bisa, tapi tidak dianjurkan
```

```python
class SimpleExample:
    def __init__(self, name):
        """
        Konstruktor: Dipanggil saat objek dibuat.
        Menyimpan nilai 'name' dan mencetak pesan pembuatan objek.
        """
        self.name = name
        print(f"Konstruktor: Objek '{self.name}' telah dibuat.")

    def __del__(self):
        """
        Destruktor: Dipanggil saat objek dihapus.
        Mencetak pesan bahwa objek sedang dihapus.
        """
        print(f"Destruktor: Objek '{self.name}' sedang dihapus.")


def main():
    print("Program dimulai.\n")

    # Membuat objek
    obj = SimpleExample("Demo")
    print("Program sedang berjalan...\n")

    # Menghapus objek
    del obj
    print("Objek telah dihapus secara eksplisit.\n")

    print("Program selesai.")


if __name__ == "__main__":
    main()
```

##Praktikum 4

```python
class FileLogger:
    def __init__(self, filename):
        """
        Konstruktor: Membuka file log untuk menulis pesan.
        Parameter:
        - filename: Nama file tempat pesan log akan ditulis.
        """
        self.filename = filename
        try:
            # Membuka file dalam mode append
            self.file = open(filename, "a")
            print(f"File '{filename}' berhasil dibuka untuk logging.")
        except Exception as e:
            print(f"Gagal membuka file '{filename}': {e}")

    def write_log(self, message):
        """
        Menulis pesan log ke dalam file.
        Parameter:
        - message: Pesan yang akan ditulis ke file.
        """
        if hasattr(self, "file") and not self.file.closed:
            self.file.write(message + "\n")
            self.file.flush()  # langsung ditulis ke disk
            print(f"Pesan log: '{message}' telah ditulis.")
        else:
            print("File tidak tersedia untuk menulis log.")

    def __del__(self):
        """
        Destruktor: Menutup file log ketika objek dihapus.
        """
        if hasattr(self, "file") and not self.file.closed:
            self.file.close()
            print(f"File '{self.filename}' telah ditutup.")


# Contoh penggunaan
if __name__ == "__main__":
    # Membuat objek logger
    logger = FileLogger("application.log")

    # Menulis log
    logger.write_log("Aplikasi dimulai.")
    logger.write_log("Melakukan operasi A...")
    logger.write_log("Operasi A selesai.")
    logger.write_log("Aplikasi akan segera selesai.")

    # Menghapus objek
    del logger

    # Jika tidak dihapus manual, __del__ akan dipanggil saat program selesai
```

##Praktikum 5

```python
class Calculator:
    def __init__(self, initial_value=0):
        """
        Konstruktor kelas Calculator.
        """
        self.value = initial_value
        print(f"Kalkulator diinisialisasi dengan nilai: {self.value}")

    def add(self, number):
        """Menambahkan nilai."""
        self.value += number
        print(f"Setelah penambahan {number}, nilai sekarang adalah: {self.value}")

    def subtract(self, number):
        """Mengurangi nilai."""
        self.value -= number
        print(f"Setelah pengurangan {number}, nilai sekarang adalah: {self.value}")

    def reset(self):
        """Mengatur ulang nilai ke 0."""
        self.value = 0
        print("Nilai telah direset ke 0.")

    def show_value(self):
        """Menampilkan nilai saat ini."""
        print(f"Nilai saat ini adalah: {self.value}")


# Contoh penggunaan
def main():
    # Objek pertama
    calc1 = Calculator(initial_value=10)
    calc1.add(5)
    calc1.subtract(3)
    calc1.show_value()

    print()  # spasi biar rapi

    # Objek kedua
    calc2 = Calculator()
    calc2.add(20)
    calc2.show_value()

    print()

    # Reset contoh
    calc1.reset()
    calc1.show_value()


if __name__ == "__main__":
    main()
```

```python
class Person:
    def __init__(self, name, age):
        """
        Konstruktor untuk menginisialisasi objek Person.
        """
        self.__name = name
        self.__age = age

    @property
    def name(self):
        """Getter untuk name"""
        return self.__name

    @name.setter
    def name(self, value):
        """Setter untuk name"""
        if not value:
            print("Nama tidak boleh kosong.")
        else:
            self.__name = value

    @property
    def age(self):
        """Getter untuk age"""
        return self.__age

    @age.setter
    def age(self, value):
        """Setter untuk age"""
        if value < 0:
            print("Umur tidak boleh negatif!")
        else:
            self.__age = value


# Contoh penggunaan
def main():
    # Membuat objek
    person = Person("Alice", 30)
    print(f"Nama: {person.name}, Umur: {person.age}")

    # Ubah data
    person.name = "Bob"
    person.age = 35
    print(f"Nama baru: {person.name}, Umur baru: {person.age}")

    # Input tidak valid
    person.age = -5  # Akan ditolak


if __name__ == "__main__":
    main()
```

```python
class Student:
    def __init__(self, name, score):
        # atribut private
        self.__name = name
        self.__score = 0
        self.__grade = ""
        self.score = score  # pakai setter biar langsung validasi

    # Getter name
    @property
    def name(self):
        return self.__name

    # Setter name
    @name.setter
    def name(self, value):
        if value:
            self.__name = value
        else:
            print("Nama tidak boleh kosong.")

    # Getter score
    @property
    def score(self):
        return self.__score

    # Setter score + validasi + update grade otomatis
    @score.setter
    def score(self, value):
        if 0 <= value <= 100:
            self.__score = value
            self.__update_grade()
        else:
            print("Score harus antara 0 - 100.")

    # Getter grade
    @property
    def grade(self):
        return self.__grade

    # Method internal untuk update grade
    def __update_grade(self):
        if 90 <= self.__score <= 100:
            self.__grade = "A"
        elif 80 <= self.__score <= 89:
            self.__grade = "B"
        elif 70 <= self.__score <= 79:
            self.__grade = "C"
        elif 60 <= self.__score <= 69:
            self.__grade = "D"
        else:
            self.__grade = "E"

    # Method show_info
    def show_info(self):
        print(f"Nama Mahasiswa: {self.__name}")
        print(f"Nilai: {self.__score}")
        print(f"Grade: {self.__grade}")

    # Destruktor
    def __del__(self):
        print(f"Data mahasiswa {self.__name} telah dihapus dari sistem.")


# ====== MAIN PROGRAM ======
def main():
    # buat objek
    mhs = Student("Bilad", 87)
    mhs.show_info()

    print("\nNilai diubah...\n")

    # ubah nilai
    mhs.score = 93
    mhs.show_info()

    # hapus objek
    del mhs


if __name__ == "__main__":
    main()
```
