## Praktikum 1: Mendefinisikan Kelas Abstrak  Sederhana



```python
# Impor komponen yang diperlukan dari modul abc
from abc import ABC, abstractmethod

# 1. Definisikan Kelas Abstrak
class KendaraanAbstrak(ABC):

    def __init__(self, merk):
        self.merk = merk
        print(f"Inisialisasi KendaraanAbstrak dengan merk: {self.merk}")

    # Metode konkret (tidak abstrak)
    def info_merk(self):
        print(f"Merk kendaraan ini adalah {self.merk}.")

    # 2. Metode abstrak
    @abstractmethod
    def start_mesin(self):
        pass

    @abstractmethod
    def stop_mesin(self):
        pass


# Kode utama
if __name__ == "__main__":
    print("Definisi Kelas Abstrak 'KendaraanAbstrak' selesai.")

    # Kelas turunan (konkret)
    class Mobil(KendaraanAbstrak):

        def start_mesin(self):
            print(f"Mesin mobil {self.merk} dinyalakan.")

        def stop_mesin(self):
            print(f"Mesin mobil {self.merk} dimatikan.")

    print("\nContoh definisi kelas anak 'Mobil' selesai.")

    # Membuat objek
    mobil_contoh = Mobil("Toyota")

    # Memanggil method
    mobil_contoh.start_mesin()
    mobil_contoh.info_merk()
    mobil_contoh.stop_mesin()
```

##  Praktikum 2: Mencoba Instansiasi Kelas    Abstrak


```python
from abc import ABC, abstractmethod

# Definisikan kelas abstrak
class MediaAbstrak(ABC):

    def __init__(self, judul):
        self.judul = judul
        print(f"Inisialisasi MediaAbstrak dengan judul: {self.judul}")

    @abstractmethod
    def play(self):
        """Metode abstrak untuk memulai pemutaran."""
        pass

    @abstractmethod
    def stop(self):
        """Metode abstrak untuk menghentikan pemutaran."""
        pass


# --- Kode Utama ---
if __name__ == "__main__":
    print("Mencoba membuat objek dari kelas abstrak MediaAbstrak...")

    # Blok try-except untuk menangkap error
    try:
        # Ini akan ERROR karena class masih abstrak
        media = MediaAbstrak("Konten Abstrak")

        # Baris ini tidak akan dijalankan
        print("Objek berhasil dibuat (SEHARUSNYA TIDAK TERJADI).")
        media.play()

    except TypeError as e:
        print("\nGAGAL membuat objek!")
        print(f"Error yang muncul (sesuai harapan): {e}")
        print("\nIni membuktikan bahwa kelas abstrak tidak bisa diinstansiasi")
        print("jika masih memiliki metode abstrak yang belum diimplementasikan.")
```

## Praktikum 3: Membuat Subclass Konkret

```python
from abc import ABC, abstractmethod
import random

# Format rupiah manual (AMAN, tanpa locale)
def format_rupiah(angka):
    return f"Rp {angka:,}".replace(",", ".")

# Kelas Abstrak
class AlatPembayaranAbstrak(ABC):

    def __init__(self, nama_metode):
        self.nama_metode = nama_metode
        print(f"Inisialisasi alat pembayaran: {self.nama_metode}")

    def info(self):
        print(f"Metode Pembayaran: {self.nama_metode}")

    @abstractmethod
    def proses_pembayaran(self, jumlah):
        pass


# Subclass 1: Kartu Kredit
class KartuKredit(AlatPembayaranAbstrak):

    def __init__(self, nomor_kartu, nama_pemilik):
        super().__init__("Kartu Kredit")
        self.nomor_kartu = nomor_kartu[-4:]
        self.nama_pemilik = nama_pemilik
        print(f" -> Kartu Kredit ************{self.nomor_kartu} ({self.nama_pemilik}) siap.")

    def proses_pembayaran(self, jumlah):
        print(f"Memproses pembayaran {format_rupiah(jumlah)} via Kartu Kredit ************{self.nomor_kartu}...")

        berhasil = random.choice([True, False])
        if berhasil:
            print("Pembayaran Kartu Kredit Berhasil.")
            return True
        else:
            print("Pembayaran Kartu Kredit Gagal.")
            return False


# Subclass 2: Dompet Digital
class DompetDigital(AlatPembayaranAbstrak):

    def __init__(self, nomor_telepon, nama_provider):
        super().__init__(f"Dompet Digital ({nama_provider})")
        self.nomor_telepon = nomor_telepon
        self._saldo = random.randint(50000, 500000)
        print(f" -> Dompet Digital {self.nomor_telepon} siap (Saldo: {format_rupiah(self._saldo)}).")

    def proses_pembayaran(self, jumlah):
        print(f"Memproses pembayaran {format_rupiah(jumlah)} via Dompet Digital {self.nomor_telepon}...")

        if jumlah <= self._saldo:
            self._saldo -= jumlah
            print("Pembayaran Berhasil.")
            print(f"Saldo tersisa: {format_rupiah(self._saldo)}")
            return True
        else:
            print("Pembayaran Gagal (Saldo tidak cukup).")
            print(f"Saldo sekarang: {format_rupiah(self._saldo)}")
            return False


# Program utama
if __name__ == "__main__":
    print("\nMembuat Objek Alat Pembayaran...")

    kartu = KartuKredit("1234567890123456", "Budi")
    gopay = DompetDigital("08123456789", "GoPay")

    print("\n--- Pembayaran Kartu Kredit ---")
    kartu.info()
    print("Status:", "Sukses" if kartu.proses_pembayaran(150000) else "Gagal")

    print("\n--- Pembayaran GoPay kecil ---")
    gopay.info()
    print("Status:", "Sukses" if gopay.proses_pembayaran(75000) else "Gagal")

    print("\n--- Pembayaran GoPay besar ---")
    print("Status:", "Sukses" if gopay.proses_pembayaran(1000000) else "Gagal")
```

## Praktikum 4: Menggunakan Kelas Abstrak  untuk Polimorfisme

```python
from abc import ABC, abstractmethod
import time

# --- Kelas Abstrak ---
class DokumenAbstrak(ABC):

    def __init__(self, nama_file):
        self.nama_file = nama_file
        print(f"Inisialisasi Dokumen: {self.nama_file}")

    def info_file(self):
        print(f"Nama File: {self.nama_file}")

    @abstractmethod
    def cetak(self):
        """Metode abstrak untuk mencetak dokumen."""
        pass

    @abstractmethod
    def simpan(self):
        """Metode abstrak untuk menyimpan dokumen."""
        pass


# --- Subclass 1: Dokumen Teks ---
class DokumenTeks(DokumenAbstrak):

    def __init__(self, nama_file, isi_teks):
        super().__init__(nama_file)
        self.isi_teks = isi_teks
        print(" -> Tipe: Dokumen Teks")

    def cetak(self):
        print(f"\nMencetak Dokumen Teks '{self.nama_file}':")
        print("=" * 20)
        print(self.isi_teks)
        print("=" * 20)

    def simpan(self):
        print(f"Menyimpan Dokumen Teks '{self.nama_file}' ke disk...")
        time.sleep(0.2)
        print(" -> Berhasil disimpan.")


# --- Subclass 2: Spreadsheet ---
class Spreadsheet(DokumenAbstrak):

    def __init__(self, nama_file, jumlah_baris, jumlah_kolom):
        super().__init__(nama_file)
        self.baris = jumlah_baris
        self.kolom = jumlah_kolom
        print(" -> Tipe: Spreadsheet")

    def cetak(self):
        print(f"\nMencetak Spreadsheet '{self.nama_file}' ({self.baris} baris x {self.kolom} kolom)...")
        print(" (Menampilkan pratinjau...)")

    def simpan(self):
        print(f"Menyimpan Spreadsheet '{self.nama_file}' ke format .xlsx...")
        time.sleep(0.3)
        print(" -> Berhasil disimpan.")


# --- Fungsi Polimorfik ---
def proses_dokumen(daftar_dokumen):
    print("\n======= MEMPROSES SEMUA DOKUMEN =======")

    for doc in daftar_dokumen:
        print(f"\n--- Memproses {type(doc).__name__}: {doc.nama_file} ---")
        try:
            doc.cetak()
            doc.simpan()
        except Exception as e:
            print(f"Error saat memproses {doc.nama_file}: {e}")

    print("\n======= SELESAI MEMPROSES DOKUMEN =======")


# --- Program Utama ---
if __name__ == "__main__":
    print("Membuat Dokumen...")

    dok1 = DokumenTeks("laporan.txt", "Ini adalah isi laporan singkat.")
    dok2 = Spreadsheet("data_penjualan.xlsx", 200, 15)
    dok3 = DokumenTeks(
        "catatan_rapat.txt",
        "Poin penting:\n- Bahas budget\n- Tentukan timeline"
    )

    koleksi_dokumen = [dok1, dok2, dok3]

    # Proses semua dokumen (polimorfisme)
    proses_dokumen(koleksi_dokumen)
```

## Praktikum 5: Kelas Abstrak dengan Metode  Konkret dan Abstrak



```python
from abc import ABC, abstractmethod
import math

# Kelas Abstrak
class BangunDatarAbstrak(ABC):

    def __init__(self, nama):
        self._nama = nama  # protected
        print(f"Inisialisasi BangunDatarAbstrak: {self._nama}")

    # Property (getter)
    @property
    def nama(self):
        return self._nama

    # Metode konkret
    def info_lengkap(self):
        print(f"\n--- Info Lengkap: {self.nama} ---")

        try:
            luas = self.hitung_luas()
            print(f"Luas     : {luas:.2f}")
        except Exception as e:
            print(f"Luas     : Error ({e})")

        try:
            keliling = self.hitung_keliling()
            print(f"Keliling : {keliling:.2f}")
        except Exception as e:
            print(f"Keliling : Error ({e})")

        print("-" * 30)

    # Metode abstrak
    @abstractmethod
    def hitung_luas(self):
        pass

    @abstractmethod
    def hitung_keliling(self):
        pass


# Subclass Konkret
class Lingkaran(BangunDatarAbstrak):

    def __init__(self, radius):
        super().__init__("Lingkaran")

        if radius < 0:
            raise ValueError("Radius tidak boleh negatif")

        self.radius = radius
        print(f" -> Lingkaran dibuat (Radius: {self.radius})")

    def hitung_luas(self):
        return math.pi * (self.radius ** 2)

    def hitung_keliling(self):
        return 2 * math.pi * self.radius


# Program utama
if __name__ == "__main__":
    print("Membuat objek Lingkaran...")

    lingkaran_A = Lingkaran(10)

    print(f"\nNama bangun (via property): {lingkaran_A.nama}")

    lingkaran_A.info_lengkap()

    # (opsional)
    # luas = lingkaran_A.hitung_luas()
    # keliling = lingkaran_A.hitung_keliling()
    # print(f"Luas langsung: {luas:.2f}, Keliling langsung: {keliling:.2f}")
```

## Praktikum 6: Properti Abstrak - Elemen Grafis

```python
from abc import ABC, abstractmethod

# Kelas Abstrak
class ElemenGrafis(ABC):

    def __init__(self, id_elemen, warna):
        print(f"[ElemenGrafis __init__] Membuat elemen dengan ID: {id_elemen}")
        self._id = id_elemen
        self.warna = warna

    # Properti konkret (getter)
    @property
    def id_elemen(self):
        print(f"[ElemenGrafis id_elemen getter] Mengakses ID: {self._id}")
        return self._id

    # --- Properti Abstrak 'posisi' ---
    @property
    @abstractmethod
    def posisi(self):
        pass

    @posisi.setter
    @abstractmethod
    def posisi(self, koordinat_baru):
        pass

    # Metode Abstrak
    @abstractmethod
    def gambar(self):
        pass

    # Metode Konkret
    def info_warna(self):
        print(f"[ElemenGrafis info_warna] Warna elemen {self.id_elemen}: {self.warna}")


# --- Subclass Konkret ---
class Kotak(ElemenGrafis):

    def __init__(self, id_elemen, warna, x=0, y=0, lebar=10, tinggi=10):
        super().__init__(id_elemen, warna)
        print(f"[Kotak __init__] Menginisialisasi Kotak ID: {self.id_elemen}")
        self._x = x
        self._y = y
        self.lebar = lebar
        self.tinggi = tinggi

    # Getter posisi
    @property
    def posisi(self):
        print(f"[Kotak posisi getter] Posisi {self.id_elemen}: ({self._x}, {self._y})")
        return (self._x, self._y)

    # Setter posisi
    @posisi.setter
    def posisi(self, koordinat_baru):
        print(f"[Kotak posisi setter] Set posisi {self.id_elemen} ke {koordinat_baru}")

        if isinstance(koordinat_baru, tuple) and len(koordinat_baru) == 2:
            self._x, self._y = koordinat_baru
            print(f" -> Berhasil diubah ke ({self._x}, {self._y})")
        else:
            print(" -> Gagal: harus tuple (x, y)")

    # Implementasi metode abstrak
    def gambar(self):
        print(f"[Kotak gambar] Kotak '{self.id_elemen}' warna {self.warna} "
              f"di ({self._x},{self._y}) ukuran {self.lebar}x{self.tinggi}")


# --- Program Utama ---
if __name__ == "__main__":
    print("Membuat objek Kotak...")

    kotak1 = Kotak("KotakA", "Merah", x=5, y=10, lebar=20)

    print("-" * 30)

    print("Mengakses ID...")
    id_ktk = kotak1.id_elemen

    print("-" * 30)

    print("Mengakses Posisi Awal...")
    pos_awal = kotak1.posisi

    print("-" * 30)

    print("Mengubah Posisi...")
    kotak1.posisi = (50, 60)

    print("-" * 30)

    print("Mengakses Posisi Baru...")
    pos_baru = kotak1.posisi

    print("-" * 30)

    print("Mencoba Set Posisi Salah...")
    kotak1.posisi = [100, 200]  # salah, bukan tuple

    print("-" * 30)

    print("Menggambar Kotak...")
    kotak1.gambar()

    print("-" * 30)

    print("Info Warna...")
    kotak1.info_warna()

    print("-" * 30)
```

```python
from abc import ABC, abstractmethod

# ===============================
# KELAS ABSTRAK
# ===============================
class Senjata(ABC):

    def __init__(self, nama):
        self.nama = nama
        print(f"Senjata '{self.nama}' dibuat")

    def info_nama(self):
        print(f"Nama Senjata: {self.nama}")

    # Method abstrak
    @abstractmethod
    def serang(self):
        pass

    # Properti abstrak (ammo / daya tahan)
    @property
    @abstractmethod
    def ammo(self):
        pass


# ===============================
# KELAS PEDANG
# ===============================
class Pedang(Senjata):

    def __init__(self, nama, panjang_bilah):
        super().__init__(nama)
        self.panjang_bilah = panjang_bilah
        self._daya_tahan = 100  # contoh tetap

    def serang(self):
        print(f"Pedang {self.nama} menebas dengan bilah sepanjang {self.panjang_bilah} cm!")

    @property
    def ammo(self):
        # Pedang tidak pakai ammo → kita gunakan daya tahan
        return self._daya_tahan


# ===============================
# KELAS PANAH
# ===============================
class Panah(Senjata):

    def __init__(self, nama, jumlah_anak_panah):
        super().__init__(nama)
        self._jumlah_anak_panah = jumlah_anak_panah

    def serang(self):
        if self._jumlah_anak_panah > 0:
            self._jumlah_anak_panah -= 1
            print(f"Panah {self.nama} melesat! Sisa anak panah: {self._jumlah_anak_panah}")
        else:
            print(f"Amunisi Panah {self.nama} habis!")

    @property
    def ammo(self):
        return self._jumlah_anak_panah

    @ammo.setter
    def ammo(self, jumlah):
        if jumlah >= 0:
            self._jumlah_anak_panah = jumlah
            print(f"Panah {self.nama} diisi ulang menjadi {self._jumlah_anak_panah}")
        else:
            print("Jumlah tidak valid")


# ===============================
# PROGRAM UTAMA
# ===============================
if __name__ == "__main__":

    print("\n=== MEMBUAT OBJEK ===")
    pedang1 = Pedang("Excalibur", 80)
    panah1 = Panah("Artemis", 3)

    print("\n=== SERANG PEDANG ===")
    pedang1.serang()

    print("\n=== SERANG PANAH ===")
    panah1.serang()
    panah1.serang()
    panah1.serang()
    panah1.serang()  # akan habis

    print("\n=== CEK PROPERTI ===")
    print(f"Daya tahan Pedang: {pedang1.ammo}")
    print(f"Ammo Panah: {panah1.ammo}")

    print("\n=== ISI ULANG PANAH ===")
    panah1.ammo = 5
    print(f"Ammo setelah isi ulang: {panah1.ammo}")

    # OPTIONAL: coba instansiasi abstract class
    print("\n=== TEST ABSTRACT CLASS ===")
    try:
        s = Senjata("SenjataAbstrak")
    except TypeError as e:
        print("Error:", e)
```
