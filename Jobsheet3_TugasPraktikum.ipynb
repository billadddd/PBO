```python
#Mencari Modus
import statistics

dadu = [1, 2, 3, 4, 5, 5, 6, 2, 3, 4, 5, 6, 6, 4, 3, 2, 1, 4, 3, 5, 6, 6, 5, 4, 5]
modus = statistics.mode(dadu)
print("Modus:", modus)
```

```python
#Deret(Mencari Presentil)
import numpy as np

deret = [9, 10, 11, 6, 8, 7, 7, 5, 4, 5]
p25 = np.percentile(deret, 25)
p75 = np.percentile(deret, 75)

print("Persentil 25:", p25)
print("Persentil 75:", p75)
```

```python
#Mencari Nilai Ujian dari (Analisis Kuartil dan Outlier)
import numpy as np

ujian = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10, 53, 63, 65, 50, 59, 62, 50, 51, 57, 60, 63, 65, 65, 53, 99]
ujian_sorted = sorted(ujian) # a. Sorting
print("a. Data urut:", ujian_sorted)

Q1 = np.percentile(ujian_sorted, 25) # b. Q1
Q3 = np.percentile(ujian_sorted, 75) # b. Q3
IQR = Q3 - Q1 # b. IQR
print("\nb. Nilai Q1:", Q1)
print("   Nilai Q3:", Q3)
print("   Nilai IQR:", IQR)

lower_bound = Q1 - (1.5 * IQR) # c. Lower Bound
upper_bound = Q3 + (1.5 * IQR) # c. Upper Bound
print("\nc. Lower Bound:", lower_bound)
print("   Upper Bound:", upper_bound)

# d. Menemukan Outlier
outliers = [x for x in ujian_sorted if x < lower_bound or x > upper_bound]
print("\nd. Nilai Outlier:", outliers)
```

```python
#Mencari Standar Devisa 
import numpy as np

# Membuat ulang data seterdasarkan nilai tengah (xi) dan frekuensinya
# 43 diulang 2x, 48 diulang 3x, dst.
data_bb = np.repeat([43, 48, 53, 58, 63, 68], [2, 3, 6, 4, 5, 2])

# Menggunakan np.std() seperti contoh di Latihan
std_dev = np.std(data_bb)

print("Standard Deviation:", std_dev)
```
